```python
"""
Developer Note:
This utility script validates that the directions loaded by our app (in EWKB format) 
are correct by converting the EWKB hex string to a Shapely geometry and printing 
its WKT (Well-Known Text) representation. Use this for quick checks and debugging.
"""

import binascii
from shapely import wkb

# Your EWKB hex string
wkb_hex = "0101000020E610000014058BE0FF7940C001B0285174524EC0"

# Convert hex to binary
wkb_bytes = binascii.unhexlify(wkb_hex)

# Parse the WKB (EWKB includes the SRID; here, it's 4326)
geom = wkb.loads(wkb_bytes)

# Print the geometry in Well-Known Text (WKT) format
print(geom)  # Example output: POINT(-50.123456789 -120.987654321)

```
